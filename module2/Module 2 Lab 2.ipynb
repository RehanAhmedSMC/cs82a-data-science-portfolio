```python
def for_adding(c, d):
    result = 0

    if c >= 0:
        for i in range(c):
            result = result + d
    else:
        for i in range(c, 0):
            result = result - d

    return result
print(for_adding(7,6))    
```

```python
def while_adding(a, b):
    result = 0
    if a >=0:
        while a > 0:
            result = result + b
            a = a - 1
    else:
        while a<0:
            result = result- b
            a=a+1
    return result
print(while_adding(7,6))  
```

**A markdown sentence explaining how the while loop terminates**

While loop terminates when the first argument reaches 0. In both cases we are moving toward 0.
