```python
Temperatures=[10,16,23,28,14,22,30]
for i in Temperatures:
    if i<15: #temperature less then 15° is Cold
        label = 'Cold'
    elif i>15 and i<=25: #temperature greater 15° and less then 25° is Warm
        label = 'Warm' 
    elif i>25: #temperature greater then 25° is hot
        label = 'Hot'
    else: #temperature that is not on the list is invalid
        label = 'Not a valid data'

    print(i, "°C:", label )    
```

Cutoffs:: temperature less then 15° is Cold, temperature greater 15° and less then 25° is Warm, temperature greater then 25° is hot, temperature that is not on the list is invalid
